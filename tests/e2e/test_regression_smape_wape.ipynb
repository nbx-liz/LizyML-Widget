```python
from sklearn.datasets import fetch_california_housing

from lizyml_widget import LizyWidget

data = fetch_california_housing(as_frame=True)
df = data.frame

w = LizyWidget()
w.load(df, target="MedHouseVal")
# P-030: smape / wape are first-class regression metrics in lizyml 0.11.
# Configure both so the Search Space metric chip group renders them and
# the learning curve metric switcher exposes them on completion.
w.set_config(
    {
        "model": {
            "name": "lgbm",
            "params": {
                "n_estimators": 50,
                "learning_rate": 0.1,
                "objective": "huber",
                "metric": ["smape", "wape"],
            },
        },
        "training": {
            "seed": 42,
            "early_stopping": {"enabled": True, "rounds": 10},
        },
        "evaluation": {"metrics": ["smape", "wape"], "params": {}},
    }
)
w
```
