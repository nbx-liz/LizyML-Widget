```python
from sklearn.datasets import load_breast_cancer

from lizyml_widget import LizyWidget

data = load_breast_cancer(as_frame=True)
df = data.frame
w = LizyWidget()
w.load(df, target="target")
# #133 acceptance test fixture: simulate a backend failure deterministically.
# Setting these traitlets directly is the same path the supervisor takes
# when an exception escapes the JobRunner, so it locks in the failed-state
# UI surface (banner + Re-run button) without depending on a brittle
# backend-rejection config.
w.status = "failed"
w.job_type = "fit"
w.job_index = 1
w.error = {
    "code": "BACKEND_ERROR",
    "message": "Synthetic failure for E2E coverage of failed-state rendering.",
}
w
```
