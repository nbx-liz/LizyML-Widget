```python
from sklearn.datasets import load_breast_cancer

from lizyml_widget import LizyWidget

data = load_breast_cancer(as_frame=True)
df = data.frame
w = LizyWidget()
w.load(df, target="target")
# Configure a long-enough tune that the E2E test has time to click Cancel
# before it finishes. n_trials=200 with the default trial budget keeps the
# job in 'running' for at least a few seconds even on fast hardware.
w.set_config(
    {
        "tuning": {
            "optuna": {
                "params": {"n_trials": 200},
                "space": {
                    "learning_rate": {"type": "float", "low": 0.001, "high": 0.5, "log": True},
                    "num_leaves": {"type": "int", "low": 8, "high": 256},
                },
            },
        },
    }
)
w
```

```python
# Kick off the tune. The widget runs it on a background thread, so this
# call returns immediately while the worker continues. The E2E test
# clicks the Cancel button mid-flight and asserts the failed-with-CANCELLED
# transition.
w.tune()
```
